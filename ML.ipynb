```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
from sklearn.datasets import load_breast_cancer
```

```python
cancer = load_breast_cancer()
```

```python
cancer.keys()
```

```python
print(cancer.DESCR)
```

```python
cancer.data
```

```python
print(cancer.target_names)
```

```python
cancer.feature_names
```

```python
dataset = pd.DataFrame(cancer.data,columns=cancer.feature_names)
```

```python
dataset.head()
```

```python
dataset['Cancer'] = cancer.target 
```

```python
dataset.info()
```

```python
dataset.describe()
##only numerical values are shown
```

```python
##missing values checking
dataset.isnull().sum()
```

```python
##EDA
##Correlation
dataset.corr()
```

```python
sns.heatmap(dataset)
plt.show()
```

```python
sns.pairplot(dataset)
plt.show()
```

```python
plt.scatter(dataset['smoothness error'],dataset['Cancer'])
plt.xlabel('Mean Area')
plt.ylabel('Cancer')
plt.show()
```

```python
sns.regplot(dataset,x=dataset['mean radius'],y=dataset['Cancer'])
plt.show()
```

```python
##Independent and Dependent Feature
X = dataset.iloc[:,:-1]
y = dataset.iloc[:,-1]
```

```python
X
```

```python
y
```

```python
##train_test_split
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.3,random_state=42)
```

```python
X_train
```

```python
X_test
```

```python
##Standardize the ddataset
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
X_train = scaler.fit_transform(X_train)
```

```python
X_test = scaler.transform(X_test)
```

```python
print(X_test)
```

```python
X_train
```

Model Training

```python
from sklearn.linear_model import LinearRegression
```

```python
regression = LinearRegression()
```

```python
regression.fit(X_train,y_train)
```

```python
##print the coefficients and the intercept
print(regression.coef_)
```

```python
print(regression.intercept_)
```

```python
print(regression.get_params())
```

```python
##prediction with test data
regression_predict = regression.predict(X_test)
```

```python
regression_predict
```

```python
##plot a scatter plot for the prediction
plt.scatter(y_test,regression_predict)
plt.show()
```

```python
residuals = y_test - regression_predict
```

```python
residuals
```

```python
##plot this residuals
sns.displot(residuals,kind='kde')
plt.show()
```

```python
##Scatter plot with respect to predictions and reresiduals
##uniform distribution
plt.scatter(regression_predict,residuals)
plt.show()
```

```python
from sklearn.metrics import mean_squared_error
from sklearn.metrics import median_absolute_error
print(median_absolute_error(y_test,regression_predict))
print(median_absolute_error(y_test,regression_predict))
print(np.sqrt(mean_squared_error(y_test,regression_predict)))
```

##R squared and adjusted R square

```python
from sklearn.metrics import r2_score
score = r2_score(y_test,regression_predict)
print(score)
```

```python
1 - (1-score)*(len(y_test)-1)/(len(y_test) - X_test.shape[1]-1)
```
